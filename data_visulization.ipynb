```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.countplot(data=df, x="diabetes")
plt.title("Diabetes Distribution")
plt.xlabel("Diabetes (0 = No, 1 = Yes)")
plt.ylabel("Count")
plt.show()
```

```python
sns.histplot(data=df, x="age", hue="diabetes", bins=30, kde=True,element="step")
plt.title("Age Distribution by Diabetes Status")
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()
```

```python
sns.boxplot(data=df, x="diabetes", y="bmi")
plt.title("BMI vs Diabetes")
plt.xlabel("Diabetes")
plt.ylabel("BMI")
plt.show()
```

```python
sns.boxplot(data=df, x="diabetes", y="HbA1c_level")
plt.title("HbA1c Level vs Diabetes")
plt.xlabel("Diabetes")
plt.ylabel("HbA1c Level")
plt.show()
```

```python
sns.boxplot(data=df, x="diabetes", y="blood_glucose_level")
plt.title("Blood Glucose Level vs Diabetes")
plt.xlabel("Diabetes")
plt.ylabel("Blood Glucose Level")
plt.show()
```

```python
sns.countplot(data=df, x="gender", hue="diabetes")
plt.title("Diabetes by Gender")
plt.show()
```

```python
sns.countplot(
    data=df,
    x="smoking_history",
    hue="diabetes"
)

plt.title("Diabetes by Smoking History")
plt.xticks(rotation=30)
plt.show()
```

```python
sns.countplot(data=df, x="heart_disease", hue="diabetes")
plt.title("Heart Disease vs Diabetes")
plt.xlabel("Heart Disease (0 = No, 1 = Yes)")
plt.show()
```

```python
numeric_cols = [
    "age",
    "hypertension",
    "heart_disease",
    "bmi",
    "HbA1c_level",
    "blood_glucose_level",
    "diabetes"
]

corr = df[numeric_cols].corr()

plt.figure(figsize=(10, 7))
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```

```python
sns.scatterplot(
    data=df,
    x="age",
    y="blood_glucose_level",
    hue="diabetes",
    alpha=0.5
)

plt.title("Age vs Blood Glucose Level")
plt.show()
```
